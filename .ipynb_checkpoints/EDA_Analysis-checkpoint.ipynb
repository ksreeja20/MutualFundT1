```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

print("Libraries Loaded Successfully")
```

```python
nav_df = pd.read_csv("data/processed/02_nav_history_cleaned.csv")

print(nav_df.shape)
nav_df.head()
```

```python
nav_df["date"] = pd.to_datetime(nav_df["date"])

daily_nav = nav_df.groupby("date")["nav"].mean().reset_index()

fig = px.line(
    daily_nav,
    x="date",
    y="nav",
    title="Average Daily NAV Trend (2022-2026)"
)

fig.write_image("reports/nav_trend.png")

fig.show()
```

```python
aum_df = pd.read_csv("data/raw/03_aum_by_fund_house.csv")

aum_df["date"] = pd.to_datetime(aum_df["date"])
aum_df["year"] = aum_df["date"].dt.year

plt.figure(figsize=(12,6))
sns.barplot(
    data=aum_df,
    x="year",
    y="aum_crore",
    hue="fund_house"
)

plt.title("AUM Growth by Fund House")

plt.savefig("reports/aum_growth.png")

plt.show()
```

```python
sip_df = pd.read_csv("data/raw/04_monthly_sip_inflows.csv")

sip_df["month"] = pd.to_datetime(sip_df["month"])

fig = px.line(
    sip_df,
    x="month",
    y="sip_inflow_crore",
    title="Monthly SIP Inflows"
)

fig.write_image("reports/sip_trend.png")
fig.show()
```

```python
txn_df = pd.read_csv(
    "data/processed/08_investor_transactions_cleaned.csv"
)

txn_df["age_group"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Age Group Distribution")
plt.ylabel("")

plt.savefig("reports/age_group_distribution.png")
plt.show()
```

```python
cat_df = pd.read_csv("data/raw/05_category_inflows.csv")

pivot_table = cat_df.pivot(
    index="category",
    columns="month",
    values="net_inflow_crore"
)

plt.figure(figsize=(14,6))
sns.heatmap(pivot_table)
plt.title("Category Inflow Heatmap")
plt.savefig("reports/category_heatmap.png")
plt.show()
```

```python
txn_df["gender"].value_counts().plot.pie(
    autopct="%1.1f%%",
    figsize=(6,6)
)

plt.title("Gender Distribution")
plt.ylabel("")
plt.show()
```

```python
state_data = txn_df.groupby("state")["amount_inr"].sum()

state_data.sort_values().tail(15).plot(
    kind="barh",
    figsize=(10,6)
)

plt.title("Top States by Investment Amount")
plt.savefig("reports/state_distribution.png")
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    data=txn_df,
    x="age_group",
    y="amount_inr"
)

plt.title("SIP Amount Distribution by Age Group")
plt.show()
```

```python
txn_df["city_tier"].value_counts().plot.pie(
    autopct="%1.1f%%",
    figsize=(6,6)
)

plt.title("T30 vs B30 Distribution")
plt.ylabel("")

plt.savefig("reports/city_tier_distribution.png")

print(nav_df.shape)
plt.show()
```

```python
folio_df = pd.read_csv("data/raw/06_industry_folio_count.csv")

plt.figure(figsize=(12,6))

plt.plot(
    folio_df["month"],
    folio_df["total_folios_crore"]
)

plt.xticks(rotation=45)

plt.title("Industry Folio Growth")
plt.ylabel("Folios (Crore)")

plt.savefig("reports/folio_growth.png")
plt.show()
```

```python
perf_df = pd.read_csv("data/raw/07_scheme_performance.csv")

numeric_cols = perf_df.select_dtypes(include="number")

plt.figure(figsize=(12,8))

sns.heatmap(
    numeric_cols.corr(),
    annot=True
)

plt.title("Scheme Performance Correlation Matrix")
plt.savefig("reports/correlation_matrix.png")
plt.show()
```

```python
hold_df = pd.read_csv("data/raw/09_portfolio_holdings.csv")

sector_data = hold_df.groupby("sector")["weight_pct"].sum().reset_index()

fig = px.pie(
    sector_data,
    names="sector",
    values="weight_pct",
    hole=0.4,
    title="Sector Allocation Across Equity Funds"
)

fig.write_image("reports/sector_allocation.png")

fig.show()
```
